```python
from pathlib import Path

# A ADAPTER selon ton Drive
PROJECT_DIR = Path('/content/drive/MyDrive/transformers')
SPLIT_DIR_IN_DRIVE = Path('/content/drive/MyDrive/transformers_data/split_out')

TRAIN_FILE = SPLIT_DIR_IN_DRIVE / 'train.npy'
TEST_FILE = SPLIT_DIR_IN_DRIVE / 'test.npy'

assert PROJECT_DIR.exists(), f'PROJECT_DIR introuvable: {PROJECT_DIR}'
assert (PROJECT_DIR / 'main.py').exists(), 'main.py introuvable dans PROJECT_DIR'
assert TRAIN_FILE.exists(), f'train.npy introuvable: {TRAIN_FILE}'
assert TEST_FILE.exists(), f'test.npy introuvable: {TEST_FILE}'

print('PROJECT_DIR =', PROJECT_DIR)
print('SPLIT_DIR   =', SPLIT_DIR_IN_DRIVE)
print('TRAIN_FILE  =', TRAIN_FILE)
print('TEST_FILE   =', TEST_FILE)
```
